# Exploratory data analysis
Revenue seasonality and category performance.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
orders = pd.read_csv(ROOT / 'data' / 'processed' / 'orders.csv', parse_dates=['order_timestamp'])
monthly = orders.query("order_status == 'Completed'").groupby('order_month', as_index=False).net_revenue.sum()
sns.lineplot(data=monthly, x='order_month', y='net_revenue', marker='o')
plt.xticks(rotation=45); plt.title('Monthly completed-order revenue'); plt.show()
```
