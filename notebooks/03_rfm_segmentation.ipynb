# RFM segmentation
Scores are generated by the reproducible processing pipeline.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
rfm = pd.read_csv(ROOT / 'data' / 'processed' / 'customer_rfm.csv')
rfm.groupby('rfm_segment').agg(customers=('customer_id','size'), revenue=('monetary','sum')).sort_values('revenue', ascending=False)
```
